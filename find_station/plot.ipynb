```python
DATAPATH = "C:\\Users\\Administrator\\Desktop\\TSC_trans\\data\\dbscan_res\\clustering_results.txt"
CENTERPATH = "C:\\Users\\Administrator\\Desktop\\TSC_trans\\data\\pam_res\\bus_stations.txt"
```

```python
with open(DATAPATH, 'r') as f:
    data = []
    for line in f:
        parts = line.strip().split(',')
        # 跳过噪点
        if parts[2] == '-1':
            continue
        data.append([parts[0], parts[1]])
with open(CENTERPATH, 'r') as f:
    centers = []
    for line in f:
        parts = line.strip().split(',')
        # 跳过第一行表头
        if parts[0] == 'Station_ID':
            continue
        centers.append([parts[1], parts[2]])
```

```python
# 绘制散点图
import matplotlib.pyplot as plt

# 提取坐标数据
data_x = [float(point[0]) for point in data]
data_y = [float(point[1]) for point in data]
center_x = [float(center[0]) for center in centers]
center_y = [float(center[1]) for center in centers]

# 设置高分辨率图片
plt.figure(figsize=(12, 10), dpi=500)

# 绘制散点和中心点
plt.scatter(data_x, data_y, c='blue', s=1, alpha=0.5, label='Data Points')
plt.scatter(center_x, center_y, c='red', s=200, marker='*', 
           edgecolors='black', linewidths=1, label='Bus Stations', zorder=100)
plt.scatter(104.06, 30.67, c = 'green', s=250, marker='X', 
            edgecolors='black', linewidths=1, label='Tianfu Square', zorder=100)

# 限制坐标轴范围
plt.xlim(103.9, 104.2)
plt.ylim(30.55, 30.8)

plt.xlabel('Longitude', fontsize=12)
plt.ylabel('Latitude', fontsize=12)
plt.title('PAM Clustering Results (Chengdu Area)', fontsize=14, fontweight='bold')
plt.legend(fontsize=10)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
